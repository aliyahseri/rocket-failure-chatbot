# Phase 1: Data Understanding

This notebook performs basic data understanding and validation of the rocket failures dataset.

```python
import pandas as pd
import numpy as np
import sys
from pathlib import Path
import os
import importlib.util

# Load config module directly from file
config_path = Path(os.path.abspath('')).parent / 'config.py'
spec = importlib.util.spec_from_file_location("config", config_path)
config = importlib.util.module_from_spec(spec)
spec.loader.exec_module(config)

CSV_PATH = config.CSV_PATH
```

```python
# Load the dataset
df = pd.read_csv(CSV_PATH)
print(f"Dataset loaded with {len(df)} records")
```

## 1. Basic Overview

```python
# Shape of the dataset
print(f"Shape: {df.shape}")
print(f"Number of records: {df.shape[0]}")
print(f"Number of features: {df.shape[1]}")
```

```python
# Data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Null values
print("\nNull Values:")
print(df.isnull().sum())
```

```python
# Duplicate check
duplicates = df.duplicated().sum()
print(f"\nDuplicate records: {duplicates}")

duplicate_ids = df['incident_id'].duplicated().sum()
print(f"Duplicate incident IDs: {duplicate_ids}")
```

```python
# First few rows
print("\nFirst 5 records:")
df.head()
```

## 2. Categorical Columns Analysis

```python
# Identify categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
print(f"Categorical columns: {categorical_cols}")
```

```python
# Value counts for categorical columns
for col in categorical_cols:
    print(f"\n{col} - Value Counts:")
    print(df[col].value_counts())
    print(f"Unique values: {df[col].nunique()}")
```

## 3. Numerical Columns Analysis

```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"Numerical columns: {numerical_cols}")
```

```python
# Descriptive statistics
print("Descriptive Statistics:")
df[numerical_cols].describe()
```

## 4. Date Range Analysis

```python
# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'])

print(f"Date range: {df['date'].min()} to {df['date'].max()}")
print(f"Span: {(df['date'].max() - df['date'].min()).days / 365.25:.1f} years")
```

## 5. Data Quality Summary

```python
print("="*50)
print("DATA QUALITY SUMMARY")
print("="*50)
print(f"Total records: {len(df)}")
print(f"Total features: {len(df.columns)}")
print(f"Missing values: {df.isnull().sum().sum()}")
print(f"Duplicate records: {duplicates}")
print(f"Date range: {df['date'].min().strftime('%Y-%m-%d')} to {df['date'].max().strftime('%Y-%m-%d')}")
print(f"\nFeature breakdown:")
print(f"  - Categorical: {len(categorical_cols)}")
print(f"  - Numerical: {len(numerical_cols)}")
```
